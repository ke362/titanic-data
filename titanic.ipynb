```python
import numpy as np
import sklearn
import matplotlib.pyplot as plt
import pandas as pd
import scipy as sp
```

```python
df=pd.read_csv("train.csv")
```

```python
#欠損値の集計
print(df.isnull().sum())
```

```python
#Ageの欠損値を中央値で置換
df["Age"]=df["Age"].fillna(df["Age"].median())
print(df["Age"])
```

```python
#df["i"]はiの列だけ表示を実行
print(df["Age"].isnull().sum())
```

```python
#列の全表示
print(df.columns)
```

```python
#columnsの各欠損値
print(df["PassengerId"].isnull().sum())
print(df["Survived"].isnull().sum())
print(df["Pclass"].isnull().sum())      #Pclass:客室のチケットの階級
print(df["Name"].isnull().sum())
print(df["Sex"].isnull().sum())
#欠損値をメジアンで補完してる
print(df["Age"].isnull().sum())
print(df["SibSp"].isnull().sum())       #SibSp      :[Siblings/兄弟姉妹]:[Spouses/配偶者]
print(df["Parch"].isnull().sum())       #Parch      :同乗してる両親・子供の数[ParentsChildrenの略称]
print(df["Ticket"].isnull().sum())
print(df["Fare"].isnull().sum())        #Fare       :旅客運賃/チケットの値段
print(df["Cabin"].isnull().sum())       #Cabin      :部屋番号
print(df["Embarked"].isnull().sum())    #Embarked   :乗客した港
```

```python
#Cabinの有効値のみ表示
#一旦保留:Cabinの欠損値が多すぎるため
print(df["Cabin"].dropna())
```

```python
#Emberkedの有効値
df["Embarked"]=df["Embarked"].dropna()
#欠損値の人の情報の開示
#df[条件]
print(df[df["Embarked"].isnull()])
#Embarkedの最頻値を出力
df["Embarked"]=df["Embarked"].mode()
#Emberkedの欠損値を最頻値で置換
```

```python
#[0]はインデックス指定、最頻値は複数ある場合(同率一位)があるため最頻値の中でも0番目の値を置換する指定
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode().iloc[0])
print(df["Embarked"])
```

```python
df.isnull().sum()
```

```python
#欠損値処理、一旦終了
#カテゴリ変数の数値化　エンコーディング:性別などの数値で表せない質的変数をコンピュータが計算しやすいように数値に変換する前処理
```

```python
#指定列の2値("Sex")のエンコーディング関数　map関数
df["Sex"]=df["Sex"].map({
    "male":0,
    "female":1
})
print(df["Sex"])                #map関数：リストやイテラブル(反復可能なオブジェクト)の全ての要素に指定した関数を適用し結果のイテレータを返す組み込み関数
                                #今回はイテラブルに該当     ＊中身の確認・利用時はfor文,list()を利用して展開する必要がある
```

```python
#Embarked(出港地)のエンコーディング
df["Embarked"]=df["Embarked"].map({
    "S":0,
    "C":1,
    "Q":2,
})
print(df["Embarked"])
```
